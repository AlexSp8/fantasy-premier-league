```python
from pathlib import Path
import sys
from IPython.display import display

import pandas as pd
import matplotlib.pyplot as plt
import seaborn

project_root = Path.cwd().parent    # cwd is notebooks/, so parent is the project root
if str(project_root) not in sys.path:
    sys.path.insert(0, str(project_root))

from notebooks.data_analysis import get_outlier_rows, get_correlation_matrix
from config.settings import FEATURES_FORM, ML_FEATURES_POS
```

```python
file_path = project_root / 'fpl_data' / 'clean' / 'players_data_features.csv'

df = pd.read_csv(file_path)
df
```

```python
outliers = get_outlier_rows(df, columns=FEATURES_FORM)
for ft in FEATURES_FORM:
    outliers_ft = outliers[outliers[ft + '_is_outlier']]
    display(ft)
    display(outliers_ft[['player_name', ft]])
    seaborn.boxplot(data=df, y=ft)
    plt.show()
```

```python
for pos, ft in ML_FEATURES_POS.items():
    df_pos = df[df['position_name'] == pos]
    corr = get_correlation_matrix(df_pos, columns=ft)
    display(corr)
    seaborn.heatmap(data=corr, annot=True)
    plt.show()
```
